```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.read_csv("df_cleaned.csv")

print("Tên các cột trong dữ liệu:")
print(df.columns)
print(df.head())

# ==========================
# 1️⃣ Biểu đồ phân loại mô hình nhà (loại nhà) và nội thất
# ==========================
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='house_type', hue='furnishing_sell')
plt.title("Phân loại mô hình nhà theo nội thất")
plt.xlabel("Loại nhà")
plt.ylabel("Số lượng")
plt.legend(title='Nội thất')
plt.show()

```

```python
# ==========================
# 2️⃣ Biểu đồ phân bố giá nhà theo loại nhà và nội thất
# ==========================
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='house_type', y='price', hue='furnishing_sell')
plt.title("Phân bố giá nhà theo loại nhà và nội thất")
plt.xlabel("Loại nhà")
plt.ylabel("Giá nhà (tỷ VNĐ)")
plt.legend(title='Nội thất')
plt.show()
```

```python
# ==========================
# 3️⃣ Biểu đồ phân loại tình trạng sở hữu (property_legal_document)
# ==========================
plt.figure(figsize=(8,8))
sns.countplot(data=df, x='property_legal_document')
plt.title("Phân loại tình trạng sở hữu / giấy tờ pháp lý")
plt.xlabel("Tình trạng pháp lý")
plt.ylabel("Số lượng")
plt.xticks(rotation=30)
plt.show()
```

```python
# ==========================
# 4️⃣ Biểu đồ phân bố giá nhà theo diện tích (size) và số tầng (floors)
# ==========================
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='size', y='price', hue='house_type', size='floors', alpha=0.7)
plt.title("Phân bố giá nhà theo diện tích và số tầng")
plt.xlabel("Diện tích (m²)")
plt.ylabel("Giá nhà (tỷ VNĐ)")
plt.show()
```
