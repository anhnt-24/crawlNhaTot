```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('df_cleaned.csv')
df.head()
```

```python
plt.scatter(df['size'], df['price'],marker='o',edgecolors='k')
plt.xlabel('Kích thước')
plt.ylabel('Giá')
plt.title('Tuo· kích thước và giá')
plt.grid(True)
plt.show()
```

```python
counts = df['house_type'].value_counts()
plt.figure(figsize=(8, 6))
plt.bar(counts.index, counts.values)
plt.xlabel('Loại nhà')
plt.ylabel('Số lượng')
plt.title('Phân bố loại hình nhà')
plt.show()

```

```python
types = df['house_type'].unique()
colors = ['skyblue', 'salmon', 'limegreen', 'red']

for i in range(len(types)):
    type_data = df[df['house_type'] == types[i]]
    plt.scatter(type_data['size'], type_data['price'], label=types[i], color=colors[i])
plt.xlabel('Kích thước')
plt.ylabel('Giá')
plt.title('Phân bố bất động sản theo kích thước, giá và loại nhà')
plt.legend()
plt.grid(True)
plt.show()
```

```python
counts = df['property_legal_document'].value_counts()
print(counts)
plt.figure(figsize=(10, 6))
plt.bar(counts.index, counts.values)
plt.xlabel('Loại tài sản')
plt.ylabel('Số lượng')
plt.title('Phân bố loại tình trạng sở hữu')
plt.xticks(rotation=45, ha='right')   # 45 hoặc 60 độ
plt.tight_layout()
plt.grid(True)
plt.show()

```

```python
plt.hist(df['size'], bins=10, color='skyblue', edgecolor='black')
plt.title("Biểu đồ Histogram")
plt.xlabel("Giá trị")
plt.ylabel("Tần suất")
plt.show()
```

```python


# Tính giá trung bình theo loại nhà
avg_price = df.groupby('house_type')['price'].mean().sort_values()
# Vẽ biểu đồ cột
plt.figure(figsize=(8,6))
plt.bar(avg_price.index, avg_price.values, color=['skyblue', 'orange', 'green', 'red'], edgecolor='black')

plt.title("Giá nhà trung bình theo loại hình", fontsize=14)
plt.xlabel("Loại nhà", fontsize=12)
plt.ylabel("Giá trung bình (tỷ đồng)", fontsize=12)
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# Tính giá trung bình theo loại nhà
avg_price = df.groupby('rooms')['price'].mean().sort_values()
# Vẽ biểu đồ cột
plt.figure(figsize=(8,6))
plt.bar(avg_price.index, avg_price.values, color=['skyblue', 'orange', 'green', 'red'], edgecolor='black')

plt.title("Giá nhà trung bình theo số phòng", fontsize=14)
plt.xlabel("Loại nhà", fontsize=12)
plt.ylabel("Giá trung bình (tỷ đồng)", fontsize=12)
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
